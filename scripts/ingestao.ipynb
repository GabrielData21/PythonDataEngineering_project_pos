```python
import duckdb
import pandas as pd
import os
from datetime import datetime
```

```python
conn = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
conn.execute("SHOW TABLES").fetchall()
```

```python
arquivo = 'z0019_2.csv'
data_ingestao = datetime.now()
df = pd.read_csv(f'../landing/{arquivo}', sep=';')

df['nome_arquivo'] = arquivo
df['data_ingestao'] = data_ingestao

df
```

```python
conn.execute("""
    CREATE TABLE IF NOT EXISTS bronze_produtos (
    ATBR VARCHAR,
    MAKTX VARCHAR,
    WERKS VARCHAR,
    MAINS VARCHAR,
    LABST VARCHAR,
    nome_arquivo VARCHAR,
    data_ingestao datetime
    )
    """)
```

```python
conn.execute("INSERT INTO bronze_produtos SELECT * FROM df")
```

```python
resultado = conn.execute("SELECT * FROM bronze_produtos").fetchdf()
resultado.head(10)
```
