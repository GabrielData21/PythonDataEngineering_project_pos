```python
import duckdb
```

```python
import os
print(os.getcwd()) 

assert os.path.exists('dados_duckdb.db'), "Banco não encontrado: o kernel não está na raiz do projeto ou a ingestão não foi executada"
conn = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
df = conn.execute("SELECT * FROM produtos").fetchdf()
df.head(10)
```

```python
df_drop = df.drop(columns=['categoria', 'id_fornecedor'])
df_drop.head(10)
```

```python
df_select = df[["id", "descricao_produto", "valor"]]
df_select.head(10)
```

```python
conn.execute("""
    CREATE TABLE IF NOT EXISTS dim_produtos (
        id_produto BIGINT,
        descricao_produto TEXT,
        valor FLOAT
        )
    """)
```

```python
conn.execute("INSERT INTO dim_produtos SELECT * FROM df_drop")
```

```python
df_dim = conn.execute("SELECT * FROM dim_produtos").fetchdf()
df_dim.head(10)
```

```python
conn.close()  # libera o arquivo do banco para os outros notebooks
```
